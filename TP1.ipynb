```python
#1 - Aquecendo os motores.
usuarios = [["Matheus", 22, "Leme", "SP"],
            ["Gabriela", 20, "Leme", "SP"],
            ["", 24, "Araras", "SP"],
            ["Gabriel", 28, "", "SP"],
            ["Juliane", 33, "Rio de janeiro", "RJ"]]
```

```python
#2 - Perfil.
def ler_usuarios():
    """
    Realiza a criação de novos dicionários
    
    Args:
    Lista: Lista a qual será lida os registros.
    
    Return:
    Lista: Lista com os devidos dicionários criados.
    """
    perfis = []
    for usuario in usuarios:
        nome = str(usuario[0])
        idade = int(usuario[1])
        cidade = str(usuario[2])
        estado = str(usuario[3])
        dicionario = {"nome": nome, "idade": idade, "localizacao": (cidade, estado)}
        perfis.append(dicionario)
    return perfis
ler_usuarios()
usuarios = ler_usuarios()
```

```python
#3 - Comparando estruturas.

#Listas: Lista é uma coleção ordenada e mutável. Permite itens duplicados. No contexto da análise de dados do INFwebNET, poderia armazenar dicionários contendo informações do usuário. 
#Tuplas: Tupla é uma coleção ordenada e imutável. Permite itens duplicados. No contexto da análise de dados do INFwebNET, poderia armazenar os dados fixos do usuário, por exemplo, uma data de nascimento.
#Dicionários: Dicionário é uma coleção ordenada e mutável. Nenhum item duplicado. No contexto da análise de dados do INFwebNET, poderia armazenar as informações dos usuários de forma mais descritiva.
```

```python
#4 - Limpando o terreno.
def limpar(perfis):
    """
    Realiza a limpeza de registros que não possuem cidade ou nome cadastrados.
    
    Args:
    Lista: Lista a qual será lida os registros.
    
    Return:
    Lista: Lista com a devida limpeza feita.
    """
    perfis_validos = []
    for perfil in perfis:
        if perfil["nome"] == "" or perfil["localizacao"][0] == "":
            continue
        perfis_validos.append(perfil)
    return perfis_validos    
limpar(usuarios)
perfis_limpos = limpar(usuarios)
```

```python
#5 - Carregando dados.
def carregar_dados(perfis_validos):
    """
    Carrega os dados do arquivo txt disponibilizado pelo professor fazendo um leitura sequencial no mesmo e adicionando em um dicionário e depois inserindo na lista principal.
    
    Args:
    Lista: Lista a qual será adicionado os registros.
    
    Return:
    Lista: Lista com os devidos dados inseridos.
    """
    with open('base_inicial.txt', 'r', encoding='utf-8') as arquivo:        
        dados = arquivo.readlines()
    for dado in range(1, len(dados)):
        dados_formatados = [dado.strip() for dado in dados[dado].split("?")]
        nome = dados_formatados[0]
        idade = dados_formatados[1]
        cidade = dados_formatados[2]
        estado = dados_formatados[3]
        amigos = dados_formatados[4:]
        dicionario = {"nome": nome, "idade": idade, "localizacao":(cidade, estado), "amigos": set(amigos)}
        perfis_validos.append(dicionario)
    print(perfis_validos)
    return perfis_validos

perfis_carregados = carregar_dados(perfis_limpos)
```

```python
#6 - Concatenando dados.
def concatena(perfis_validos):
    """
    Realiza a inserção de amigos para os usuários que foram inseridos no início do programa e estavam sem amigos e também nome aos usuários sem nome
    
    Args:
    Lista: Lista a qual será atualizado os registros.
    
    Return:
    Lista: Lista com os devidos dados atualizados.
    """
    for i in range(0, len(perfis_validos)):
        if len(perfis_validos[i]) < 4:
            perfis_validos[i].update({"amigos": set(["Aline", "Lucas"])})
        
        if not perfis_validos[i].get("nome"):
            perfis_validos[i]["nome"] = "Desconhecido"
    
    with open('rede_INFNET.txt', 'w', encoding="utf-8") as arquivos:
        for perfis in perfis_validos:
            arquivos.write(str(perfis) + '\n')

    return perfis_validos

perfis_concatenados = concatena(perfis_carregados)
```

```python
#7 - Adicionando amigos.
def adiciona(perfis_validos):
    """
    Adicionando um novo amigo para um usuário específico.
    
    Args:
    Lista: Lista a qual será adicionado o amigo.
    
    Return:
    Lista: Lista com os devidos dados alterados.
    """
    perfis_validos[0]["amigos"].add("Gabriela")

    #Atualizando o arquivo txt.
    with open('rede_INFNET.txt', 'w', encoding="utf-8") as arquivos:
        for perfis in perfis_validos:
            nome = perfis["nome"]
            idade = perfis["idade"]
            localizacao = (perfis["localizacao"])  # Concatenando a localização
            amigos = (perfis["amigos"]) 
            linha = f"{nome} - {idade} - {localizacao} - {amigos}\n"
            dados = arquivos.write(linha)
    return perfis_validos

amigos_adicionados = adiciona(perfis_concatenados)
```

```python
#8 - Verificando conexões
def verificaConexoes(perfis_validos):
    """
    Verifica se um determinado usuário é popular ou não.
    
    Args:
    Lista: Lista a qual será lido os registros.
    
    Return:
    Lista: Impressão indicando se o determinado usuário é popular ou não.
    """
    contador = 0
    try:
        usuario = str(input("Qual usuário deseja descobrir se foi adicionado por mais de 4(quatro) usuários? ")).strip()
        for perfil in perfis_validos:
            if "amigos" in perfil and usuario in perfil["amigos"]:
                contador+=1
                
        if contador > 4:
            print(f"O usuário {usuario} é popular, tem {contador} amigos!")
        else:
            print(f"O usuário {usuario} não é popular, tem apenas {contador} amigos!")
    except ValueError:
        print("Entrada inválida!")
        
verificaConexoes(amigos_adicionados)
```

```python
#9 - Amigos em comum.
from random import randint

def amigosEmComum(perfis_validos):
    """
    Leitura da lista de usuários para pegar dois usuários em específico usando a biblioteca random e retornar os amigos de cada um para fazer a comparação e descobrir os amigos em comum.
    
    Args:
    Lista: Lista a qual será feito a leitura os registros.
    
    Return:
    Lista: Impressão dos amigos em comum, a posição dos dois usuários que foram escolhidos através da biblioteca random e o nome de cada um.
    """
    primeiro_usuario = set()
    segundo_usuario = set()
    primeiro_aleatorio = randint(0, len(perfis_validos))
    segundo_aleatorio = randint(0, len(perfis_validos))
    
    if "amigos" in perfis_validos[primeiro_aleatorio]:
        for amigo in perfis_validos[primeiro_aleatorio]["amigos"]:
            primeiro_usuario.add(amigo)
            
    if "amigos" in perfis_validos[segundo_aleatorio]:
        for amigo in perfis_validos[segundo_aleatorio]["amigos"]:
            segundo_usuario.add(amigo)     

    comuns = primeiro_usuario.intersection(segundo_usuario)
    nomePrimeiro = perfis_validos[primeiro_aleatorio]["nome"]
    nomeSegundo = perfis_validos[segundo_aleatorio]["nome"]
    print(f"O usuário {nomePrimeiro} tem como amigo os seguintes usuários: {primeiro_usuario}")   
    print(f"O usuário {nomeSegundo} tem como amigo os seguintes usuários: {segundo_usuario}") 
    if len(comuns) < 1:
        print("Nenhum amigo em comum!")
    else:
        print(f"Amigos em comuns: {comuns}") 
    return primeiro_usuario, segundo_usuario, nomePrimeiro, nomeSegundo  
    
#amigosEmComum(amigos_adicionados)
primeiroUsuario, segundoUsuario, nomePrimeiro, nomeSegundo = amigosEmComum(amigos_adicionados)
```

```python
#10 - Conexões exclusivas.
def conexoesExclusiva(primeiro_usuario, segundo_usuario, nomePrimeiro, nomeSegundo):
    """
    Leitura da lista de usuários para pegar dois usuários em específico usando a biblioteca random e retornar os amigos de cada um para fazer a comparação e descobrir os amigos em comum.
    
    Args:
    Lista: Um retorno da função amigosEmComum com a posição dos dois usuários que foram escolhidos através da biblioteca random e o nome de cada um.
    
    Return:
    Lista: Impressão com os amigos diferentes.
    """
    diferenca = primeiro_usuario.difference(segundo_usuario)
    print(f"O usuário {nomePrimeiro} tem como amigo os seguintes usuários: {primeiro_usuario}")   
    print(f"O usuário {nomeSegundo} tem como amigo os seguintes usuários: {segundo_usuario}") 
    print(f"Amigos diferentes: {diferenca}")     

conexoesExclusiva(primeiroUsuario, segundoUsuario, nomePrimeiro, nomeSegundo)
```

```python
#11 - Removendo conexões.
def removeConexoes(perfis_validos):
    """
    Removendo um amigo específico a partir da escolha do usuário através dos prompts.
    
    Args:
    Lista:  Lista a qual será feito a removido algum amigo.
    
    Return:
    Lista: Lista com os devidos dados removidos.
    """
    for c in range(0, len(perfis_validos)):
        nome = str(perfis_validos[c]["nome"])
        idade = int(perfis_validos[c]["idade"])
        localizacao = perfis_validos[c]["localizacao"]
        amigos = perfis_validos[c]["amigos"]
        print(f"{c+1} - {nome} - {idade} - {localizacao} - {amigos}")
    
    try:
        usuario = int(input("De qual usuário deseja remover um amigo? (Escolha a posição)"))
        remover = str(input("Qual o nome do amigo que deseja remover desse usuário? "))
        
        if (usuario-1) >= len(perfis_validos) or (usuario-1) < 0:
            print("Usuário inválido, tente novamente!")
            return perfis_validos
        
        if remover not in perfis_validos[usuario-1]["amigos"]:
            print("Amigo inválido, tente novamente!")
            return perfis_validos     

        perfis_validos[usuario - 1]["amigos"].remove(remover)
        print("Removido com sucesso!\n")                
        
    except ValueError:
        print("Digite um valor válido!")
    
    return perfis_validos
    
removeAmigos = removeConexoes(amigos_adicionados)
```

```python
#12 - Salvando o progresso.
def salvandoProgresso(perfis_validos):
    """
    Criando um novo arquivo txt com as alterações a partir de uma leitura sequencial da lista de usuários.
    
    Args:
    Lista: Lista a qual será feito a leitura para a inserção no txt.
    
    Return:
    Lista: Não necessita de retorno.
    """
    with open('rede_INFNET_atualizado.txt', 'w', encoding="utf-8") as arquivos:
        for perfis in perfis_validos:
            arquivos.write(str(perfis) + '\n')

salvandoProgresso(removeAmigos)
```

```python
#13 - Listando usuários.

def listandoUsuarios():
    """
    Exibindo os usuários que estão presentes no arquivo txt a partir de uma leitura sequencial.
    
    Args:
    Lista: Nenhum.
    
    Return:
    Lista: Impressão dos nomes dos usuários.
    """
    with open('rede_INFNET.txt', 'r', encoding="UTF-8") as arquivos:
        dados = arquivos.readlines()
        
    for linha in dados:
        linha = linha.split("-")
        print(linha[0])
        
listandoUsuarios()
```

```python
#14 -  Quantidade de Amigos.
def quantidadeAmigos():
    """
    Exibindo o nome e a quantidade de amigos dos usuários que estão presentes no arquivo txt a partir de uma leitura sequencial.
    
    Args:
    Lista: Nenhum.
    
    Return:
    Lista: Impressão dos nomes e a quantidade de amigos.
    """
    with open('rede_INFNET.txt', 'r', encoding="UTF-8") as arquivos:
        dados = arquivos.readlines()
        
    for linha in dados:
        linha = linha.split("-")
        nome = linha[0].strip()
        amigos = linha[3].split(",")
        quantidade = len(amigos)
        amigos = ", ".join(amigos)      
        print(f"O usuário {nome} tem como amigo {quantidade} pessoas, que são: {amigos}")   
    
quantidadeAmigos()
```

```python
# Usuários Mais Populares 
def maisPopulares():
    """
    Exibindo os cinco usuários que mais tem amigos no INFwebNET.
    
    Args:
    Lista: Nenhum.
    
    Return:
    Lista: Impressão dos nomes e a quantidade de amigos dos usuários que mais possuem amigos.
    """
    lista = []
    usuarios = []
    
    with open('rede_INFNET.txt', 'r', encoding="UTF-8") as arquivos:
        dados = arquivos.readlines()
    
    #Grava nome dos usuários
    for linha in dados:
        linha = linha.strip().split(" - ")
        nome = linha[0].strip()
        usuarios.append(nome)

    #Percorre as linhas e compara para ver se o usuário está presente na lista de amigos desse usuário
    for usuario in usuarios:
        contador = 0 #Limpa o retorno
        for linha in dados:
            linha = linha.strip().split(" - ")
            amigos = linha[3].strip("()").replace("'", "").split(", ")  #Formata a lista de amigos do usuário

            if usuario in amigos:
                contador += 1
        
        lista.append((usuario, contador))

    lista.sort(key=lambda x: x[1], reverse=True)
  
    print("Pessoas mais populares")
    for pessoa in range(0, 5):
        print(f"Usuário: {lista[pessoa][0]} com um total de {lista[pessoa][1]} amigos")

maisPopulares()
```
